```python
def fatorial(numero):
    resultado = 1

    for i in range(1, numero + 1):
        resultado = resultado * i

    return resultado


numero = int(input("Digite um número: "))

if numero >= 0:
    print("Fatorial:", fatorial(numero))
else:
    print("Digite um número positivo.")
```

```python
def palindromo(texto):
    texto = texto.lower()

    texto_novo = ""

    for letra in texto:
        if letra.isalnum():
            texto_novo = texto_novo + letra

    invertido = ""

    for i in range(len(texto_novo) - 1, -1, -1):
        invertido = invertido + texto_novo[i]

    if texto_novo == invertido:
        return True
    else:
        return False


texto = input("Digite uma palavra ou frase: ")

if palindromo(texto):
    print("É um palíndromo!")
else:
    print("Não é um palíndromo!")
```

```python
numero = int(input("Digite um número positivo: "))

soma = 0

while numero > 0:
    digito = numero % 10
    soma = soma + digito
    numero = numero // 10

print("Soma dos dígitos:", soma)
```

```python
def primo(numero):

    if numero < 2:
        return False

    for i in range(2, numero):
        if numero % i == 0:
            return False

    return True


numero = int(input("Digite um número: "))

if primo(numero):
    print("O número é primo.")
else:
    print("O número não é primo.")
```

```python
def calcular_imc(peso, altura):
    imc = peso / (altura * altura)

    return imc


peso = float(input("Digite seu peso: "))
altura = float(input("Digite sua altura: "))

imc = calcular_imc(peso, altura)

print("Seu IMC é:", imc)

if imc < 18.5:
    print("Abaixo do peso")
elif imc < 25:
    print("Peso normal")
elif imc < 30:
    print("Sobrepeso")
else:
    print("Obesidade")
```

```python
print("1 - Celsius para Fahrenheit")
print("2 - Fahrenheit para Celsius")

opcao = int(input("Escolha uma opção: "))

temperatura = float(input("Digite a temperatura: "))

if opcao == 1:
    resultado = (temperatura * 9 / 5) + 32

    print("Temperatura em Fahrenheit:", resultado)

elif opcao == 2:
    resultado = (temperatura - 32) * 5 / 9

    print("Temperatura em Celsius:", resultado)

else:
    print("Opção inválida!")
```

```python
numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

print("+ = Adição")
print("- = Subtração")
print("* = Multiplicação")
print("/ = Divisão")

operacao = input("Escolha a operação: ")

if operacao == "+":
    resultado = numero1 + numero2
    print("Resultado:", resultado)

elif operacao == "-":
    resultado = numero1 - numero2
    print("Resultado:", resultado)

elif operacao == "*":
    resultado = numero1 * numero2
    print("Resultado:", resultado)

elif operacao == "/":
    if numero2 != 0:
        resultado = numero1 / numero2
        print("Resultado:", resultado)
    else:
        print("Não é possível dividir por zero.")

else:
    print("Operação inválida!")
```
