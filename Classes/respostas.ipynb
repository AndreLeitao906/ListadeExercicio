```python
class Circulo:
  def __init__(self, raio):
    self.raio = raio

  def calcular_area(self):
    area = 3.14 * (self.raio ** 2)
    return area

circulo1 = Circulo(5)

print("Questão 1")
print("Área do círculo: ", circulo1.calcular_area())
```

```python
class Livro:
  def __init__(self, titulo, autor):
    self.titulo = titulo
    self.autor = autor

  def detalhes(self):
    return "Título: " + self.titulo + " | Autor: " + self.autor

livro1 = Livro("Harry Potter", "J.K. Rowling")

print(livro1.detalhes())

```

```python
class Retangulo:
  def __init__(self, base, altura):
    self.base = base
    self.altura = altura

  def calcular_area(self):
    area = self.base * self.altura
    return area

retangulo1 = Retangulo(10, 5)

print("Área do retângulo:", retangulo1.calcular_area())
```

```python
class ContaBancaria:
  def __init__(self, saldo, titular):
    self.saldo = saldo
    self.titular = titular

  def depositar(self, valor):
    self.saldo = self.saldo + valor
  def sacar(self, valor):
      if valor <= self.saldo:
        self.saldo = self.saldo - valor
      else:
        print("Saldo insuficiente")

conta1 = ContaBancaria(1000, "André")

print("Titular:", conta1.titular)
print("Saldo inicial:", conta1.saldo)
conta1.depositar(500)
print("Saldo após depósito:", conta1.saldo)
conta1.sacar(200)
print("Saldo após saque:", conta1.saldo)
```

```python
class Pessoa:
  def __init__(self, nome, idade):
    self.nome = nome
    self.idade = idade

  def falar(self):
    print("Olá, meu nome é", self.nome)

pessoa1 = Pessoa("André", 28)

pessoa1.falar()
```

```python
class Produto:
  def __init__(self, nome, preco, quantidade):
    self.nome = nome
    self.preco = preco
    self.quantidade = quantidade

  def calcular_total(self):
    total = self.preco * self.quantidade
    return total

produto1 = Produto("Teclado", 100, 3)

print("Produto:", produto1.nome)
print("Valor total:", produto1.calcular_total())
```

```python
class Carro:
  def __init__(self, marca, modelo, ano):
    self.marca = marca
    self.modelo = modelo
    self.ano = ano

  def detalhes(self):
    return "Marca: " + self.marca + " | Modelo: " + self.modelo + " | Ano: " + str(self.ano)

carro1 = Carro("Toyota", "Corolla", 2024)

print(carro1.detalhes())
```

```python
class Aluno:
  def __init__(self, nome, notas):
    self.nome = nome
    self.notas = notas

  def calcular_media(self):
    media = sum(self.notas) / len(self.notas)
    return media

aluno1 = Aluno("André", [8, 7, 9])

print("Aluno:", aluno1.nome)
print("Média:", aluno1.calcular_media())
```

```python
class Triangulo:
  def __init__(self, lado1, lado2, lado3):
    self.lado1 = lado1
    self.lado2 = lado2
    self.lado3 = lado3
  def calcular_perimetro(self):
    perimetro = self.lado1 + self.lado2 + self.lado3
    return perimetro

triangulo1 = Triangulo(5, 6, 7)

print("Perímetro do triângulo:", triangulo1.calcular_perimetro())
```

```python
class Funcionario:
  def __init__(self, nome, salario, cargo):
    self.nome = nome
    self.salario = salario
    self.cargo = cargo

  def aumentar_salario(self, percentual):
    aumento = self.salario * percentual / 100
    self.salario = self.salario + aumento

funcionario1 = Funcionario("André", 2000, "Programador")

print("Nome:", funcionario1.nome)
print("Cargo:", funcionario1.cargo)
print("Salário inicial:", funcionario1.salario)
funcionario1.aumentar_salario(10)
```
